```python
df = data[data.columns[4:21]]

# pega só as colunas Likert (valores no formato "5 - Concordo totalmente")
cols = [c for c in df.columns if df[c].astype(str).str.match(r"^[1-5] - ").all()]
likert = df[cols].apply(lambda s: s.astype(str).str[0].astype(int))
likert.columns = [f"Q{i}" for i in range(1, len(cols) + 1)]

# percentual por nível (1 a 5), zerando níveis sem resposta
pct = (likert.apply(lambda s: s.value_counts(normalize=True))
       .reindex(range(1, 6)).fillna(0).T * 100)

cores = ["#00FFDD", "#C5EC00", "#E5E5E5", "#FFC74E", "#CA00CE"]  # 1 -> 5

rotulos = {
    1: "1 - Discordo totalmente",
    2: "2 - Discordo parcialmente",
    3: "3 - Nem concordo nem discordo",
    4: "4 - Concordo parcialmente",
    5: "5 - Concordo totalmente",
}

fig, ax = plt.subplots(figsize=(10, 0.4 * len(pct) + 1.8))
esq = pd.Series(0.0, index=pct.index)
for nivel, cor in zip(range(1, 6), cores):
    ax.barh(pct.index, pct[nivel], left=esq, color=cor,
            edgecolor="white", label=rotulos[nivel]) 
    esq += pct[nivel]

ax.legend(ncol=3, loc="lower center", bbox_to_anchor=(0.5, 1.18),
          frameon=False, fontsize=8,
          handlelength=1.2, handletextpad=0.5, columnspacing=1.0)

ax.set_xlim(0, 100)
ax.invert_yaxis()                      # Q1 no topo
ax.set_xlabel("Percentual de respondentes (%)")

# eixo superior espelhado (100 -> 0)
ax.secondary_xaxis("top", functions=(lambda x: 100 - x, lambda x: 100 - x))

ax.legend(ncol=5, loc="lower center", bbox_to_anchor=(0.5, 1.15), frameon=False)
for lado in ["top", "right", "left"]:
    ax.spines[lado].set_visible(False)
ax.tick_params(left=False)
plt.bbox_inches="tight"
#fig.savefig("likert.png", dpi=300, bbox_inches="tight")
plt.show()
```
